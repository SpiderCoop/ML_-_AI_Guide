## Descripción

Las **redes neuronales artificiales (RNA)** son modelos computacionales inspirados en la estructura y funcionamiento del cerebro biológico. Están compuestas por unidades simples llamadas *neuronas* conectadas entre sí, organizadas en capas que permiten procesar información de manera jerárquica.

Una red neuronal artificial es un sistema de aprendizaje automático capaz de aprender patrones complejos en datos mediante el ajuste iterativo de los pesos de conexiones entre neuronas. Son especialmente útiles en tareas donde las relaciones entre variables son no lineales y complejas.

En este notebook exploraremos:
- La estructura básica de una neurona artificial y cómo funciona la propagación de información
- La arquitectura de un perceptrón multicapa (MLP)
- El proceso de entrenamiento mediante retropropagación (backpropagation)
- Una implementación práctica desde cero con NumPy
- Aplicaciones comunes en clasificación y regresión

## Objetivo del Modelo

- **Tipo de modelo:** Modelo de aprendizaje supervisado no lineal
- **Problemas que resuelve:** 
  - Clasificación multiclase y binaria con límites de decisión no lineales
  - Regresión con relaciones complejas entre características y target
  - Extracción de características jerárquicas
- **Ventaja pedagógica:** Fundamentar la comprensión de arquitecturas neuronales más complejas (CNN, RNN, Transformers)

## Estructura del Modelo

Una red neuronal típica consta de:

1. **Capa de Entrada:** recibe los datos originales de dimensión $n_{features}$
2. **Capas Ocultas:** una o más capas con $n_h$ neuronas cada una, transforman la representación de los datos
3. **Capa de Salida:** produce predicciones de dimensión $n_{classes}$ (para clasificación) o $1$ (para regresión)
4. **Función de Activación:** introduce no linealidad (ReLU, Sigmoid, Tanh, etc.)
5. **Función de Pérdida:** mide el error del modelo (Cross-Entropy, MSE, etc.)
6. **Optimizador:** actualiza los pesos para minimizar la pérdida (SGD, Adam, etc.)

## Modelo Matemático

### Neurona Artificial (Perceptrón)

Una neurona individual recibe múltiples entradas $x_1, x_2, \ldots, x_n$ ponderadas por pesos $w_1, w_2, \ldots, w_n$ y un sesgo (bias) $b$. La salida se calcula como:

$$z = w_1 x_1 + w_2 x_2 + \cdots + w_n x_n + b = \mathbf{w}^T \mathbf{x} + b$$

Luego se aplica una función de activación $\sigma(\cdot)$ para introducir no linealidad:

$$a = \sigma(z)$$

### Funciones de Activación Comunes

1. **ReLU (Rectified Linear Unit):**
$$\sigma(z) = \max(0, z)$$

2. **Sigmoid:**
$$\sigma(z) = \frac{1}{1 + e^{-z}}$$

3. **Tanh:**
$$\sigma(z) = \frac{e^z - e^{-z}}{e^z + e^{-z}}$$

### Propagación Hacia Adelante (Forward Pass)

Para una red de $L$ capas, la propagación es:

$$\mathbf{z}^{[l]} = \mathbf{W}^{[l]} \mathbf{a}^{[l-1]} + \mathbf{b}^{[l]}$$
$$\mathbf{a}^{[l]} = \sigma^{[l]}(\mathbf{z}^{[l]})$$

donde $\mathbf{a}^{[l]}$ es la activación de la capa $l$, $\mathbf{W}^{[l]}$ son los pesos y $\mathbf{b}^{[l]}$ es el sesgo.

### Función de Pérdida

Para clasificación, se usa **Cross-Entropy**:
$$\mathcal{L} = -\frac{1}{m} \sum_{i=1}^{m} \left[ \mathbf{y}^{(i)} \log(\hat{\mathbf{y}}^{(i)}) + (1 - \mathbf{y}^{(i)}) \log(1 - \hat{\mathbf{y}}^{(i)}) \right]$$

Para regresión, se usa **Error Cuadrático Medio (MSE)**:
$$\mathcal{L} = \frac{1}{m} \sum_{i=1}^{m} (y^{(i)} - \hat{y}^{(i)})^2$$

### Retropropagación (Backpropagation)

El gradiente de la pérdida con respecto a los pesos se calcula mediante la regla de la cadena:

$$\frac{\partial \mathcal{L}}{\partial \mathbf{W}^{[l]}} = \frac{\partial \mathcal{L}}{\partial \mathbf{a}^{[l]}} \cdot \frac{\partial \mathbf{a}^{[l]}}{\partial \mathbf{z}^{[l]}} \cdot \frac{\partial \mathbf{z}^{[l]}}{\partial \mathbf{W}^{[l]}}$$

Los pesos se actualizan mediante descenso de gradiente:

$$\mathbf{W}^{[l]} \leftarrow \mathbf{W}^{[l]} - \eta \frac{\partial \mathcal{L}}{\partial \mathbf{W}^{[l]}}$$

donde $\eta$ es la tasa de aprendizaje.

## Ventajas y Desventajas

### Ventajas
- **Capacidad de representación:** pueden aproximar funciones complejas y no lineales
- **Versatilidad:** aplicables a múltiples dominios (visión, lenguaje natural, series de tiempo)
- **Escalabilidad:** funcionan bien con grandes volúmenes de datos
- **Aprendizaje de características:** capas ocultas aprenden representaciones intermedias útiles
- **Menos ingeniería de features:** requieren menos preprocesamiento manual que modelos clásicos

### Desventajas
- **Caja negra:** difícil interpretar qué aprende la red internamente
- **Requiere muchos datos:** generalmente necesitan cantidades grandes para entrenar bien
- **Computacionalmente costosa:** entrenamiento puede ser lento, especialmente con deep networks
- **Overfitting:** propenso al sobreajuste sin regularización adecuada
- **Hiperparámetros:** arquitectura, tasa de aprendizaje, número de capas requieren ajuste cuidadoso
- **Inicialización sensible:** los pesos iniciales pueden afectar los resultados finales

## Implementación: Red Neuronal desde Cero

````python
# Celda 2: Definir la clase de Red Neuronal Multicapa (MLP)

class SimpleNeuralNetwork:
    """
    Red neuronal multicapa simple con backpropagation.
    """
    
    def __init__(self, layer_sizes, activation='relu', learning_rate=0.01, random_seed=42):
        """
        Parámetros:
        - layer_sizes: list de int con el número de neuronas en cada capa
        - activation: función de activación ('relu', 'sigmoid', 'tanh')
        - learning_rate: tasa de aprendizaje para descenso de gradiente
        - random_seed: semilla para reproducibilidad
        """
        np.random.seed(random_seed)
        self.layer_sizes = layer_sizes
        self.activation_type = activation
        self.learning_rate = learning_rate
        self.L = len(layer_sizes) - 1  # Número de capas de pesos
        
        # Inicializar pesos y sesgos
        self.weights = []
        self.biases = []
        
        for l in range(self.L):
            # Inicialización He para ReLU, Xavier para otras
            scale = np.sqrt(2 / layer_sizes[l]) if activation == 'relu' else np.sqrt(1 / layer_sizes[l])
            W = np.random.randn(layer_sizes[l], layer_sizes[l+1]) * scale
            b = np.zeros((1, layer_sizes[l+1]))
            self.weights.append(W)
            self.biases.append(b)
        
        self.cache = {}  # Para almacenar valores durante forward pass
        self.training_loss = []
    
    def activation(self, Z):
        """Función de activación."""
        if self.activation_type == 'relu':
            return np.maximum(0, Z)
        elif self.activation_type == 'sigmoid':
            return 1 / (1 + np.exp(-np.clip(Z, -500, 500)))  # Clip para evitar overflow
        elif self.activation_type == 'tanh':
            return np.tanh(Z)
    
    def activation_derivative(self, A, Z):
        """Derivada de la función de activación."""
        if self.activation_type == 'relu':
            return (Z > 0).astype(float)
        elif self.activation_type == 'sigmoid':
            return A * (1 - A)
        elif self.activation_type == 'tanh':
            return 1 - A**2
    
    def forward(self, X):
        """Propagación hacia adelante."""
        A = X
        self.cache = {'A0': A}
        
        for l in range(self.L):
            Z = np.dot(A, self.weights[l]) + self.biases[l]
            if l < self.L - 1:  # Aplicar activación a todas menos a la última
                A = self.activation(Z)
            else:  # Última capa: sigmoid para clasificación binaria
                A = 1 / (1 + np.exp(-np.clip(Z, -500, 500)))
            
            self.cache[f'Z{l+1}'] = Z
            self.cache[f'A{l+1}'] = A
        
        return A
    
    def compute_loss(self, Y_pred, Y_true):
        """Calcular pérdida (Cross-Entropy para clasificación)."""
        m = Y_true.shape[0]
        epsilon = 1e-8
        Y_pred = np.clip(Y_pred, epsilon, 1 - epsilon)
        loss = -np.mean(Y_true * np.log(Y_pred) + (1 - Y_true) * np.log(1 - Y_pred))
        return loss
    
    def backward(self, Y_true):
        """Retropropagación para calcular gradientes."""
        m = Y_true.shape[0]
        grads_W = []
        grads_b = []
        
        # Iniciar con error de la capa de salida
        dA = -(Y_true / np.clip(self.cache[f'A{self.L}'], 1e-8, 1) - 
               (1 - Y_true) / np.clip(1 - self.cache[f'A{self.L}'], 1e-8, 1))
        
        for l in range(self.L - 1, -1, -1):
            Z = self.cache[f'Z{l+1}']
            A_prev = self.cache[f'A{l}']
            
            if l == self.L - 1:  # Última capa con sigmoid
                dZ = dA * (1 / (1 + np.exp(-np.clip(Z, -500, 500)))) * (1 - 1 / (1 + np.exp(-np.clip(Z, -500, 500))))
            else:
                dZ = dA * self.activation_derivative(self.cache[f'A{l+1}'], Z)
            
            dW = np.dot(A_prev.T, dZ) / m
            db = np.sum(dZ, axis=0, keepdims=True) / m
            dA = np.dot(dZ, self.weights[l].T)
            
            grads_W.insert(0, dW)
            grads_b.insert(0, db)
        
        return grads_W, grads_b
    
    def update_parameters(self, grads_W, grads_b):
        """Actualizar pesos y sesgos."""
        for l in range(self.L):
            self.weights[l] -= self.learning_rate * grads_W[l]
            self.biases[l] -= self.learning_rate * grads_b[l]
    
    def fit(self, X, Y, epochs=100, batch_size=32, verbose=True):
        """Entrenar la red neuronal."""
        for epoch in range(epochs):
            # Shuffle datos
            indices = np.random.permutation(X.shape[0])
            X_shuffled = X[indices]
            Y_shuffled = Y[indices]
            
            epoch_loss = 0
            num_batches = 0
            
            # Mini-batch gradient descent
            for i in range(0, X.shape[0], batch_size):
                X_batch = X_shuffled[i:i+batch_size]
                Y_batch = Y_shuffled[i:i+batch_size]
                
                # Forward pass
                Y_pred = self.forward(X_batch)
                loss = self.compute_loss(Y_pred, Y_batch)
                epoch_loss += loss
                num_batches += 1
                
                # Backward pass
                grads_W, grads_b = self.backward(Y_batch)
                
                # Actualizar parámetros
                self.update_parameters(grads_W, grads_b)
            
            epoch_loss /= num_batches
            self.training_loss.append(epoch_loss)
            
            if verbose and (epoch + 1) % 20 == 0:
                print(f"Epoch {epoch+1}/{epochs}, Loss: {epoch_loss:.4f}")
    
    def predict(self, X):
        """Realizar predicciones."""
        Y_pred = self.forward(X)
        return (Y_pred > 0.5).astype(int).flatten()
    
    def predict_proba(self, X):
        """Obtener probabilidades."""
        return self.forward(X)

print("Clase SimpleNeuralNetwork definida correctamente.")
````

````python
# Celda 3: Crear dataset sintético para clasificación

# Usar make_moons para un problema no linealmente separable
X, y = make_moons(n_samples=300, noise=0.1, random_state=RANDOM_SEED)

# Normalizar características
scaler = StandardScaler()
X = scaler.fit_transform(X)

# Remodelar y como matriz columna
y = y.reshape(-1, 1)

print(f"Shape de X: {X.shape}")
print(f"Shape de y: {y.shape}")
print(f"Distribución de clases: {np.bincount(y.flatten())}")

# Visualizar datos
plt.figure(figsize=(8, 6))
plt.scatter(X[y.flatten()==0, 0], X[y.flatten()==0, 1], label='Clase 0', alpha=0.7)
plt.scatter(X[y.flatten()==1, 0], X[y.flatten()==1, 1], label='Clase 1', alpha=0.7)
plt.xlabel('Característica 1')
plt.ylabel('Característica 2')
plt.title('Dataset: Make Moons')
plt.legend()
plt.grid()
plt.show()
````

````python
# Celda 4: Dividir datos en train/test

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=RANDOM_SEED
)

print(f"Training set size: {X_train.shape[0]}")
print(f"Test set size: {X_test.shape[0]}")
print(f"Training set shape: {X_train.shape}")
print(f"Test set shape: {X_test.shape}")
````

````python
# Celda 5: Entrenar la red neuronal

# Definir arquitectura: 2 entrada -> 8 oculta -> 1 salida
layer_sizes = [2, 8, 1]

# Crear e instanciar la red
nn = SimpleNeuralNetwork(
    layer_sizes=layer_sizes,
    activation='relu',
    learning_rate=0.01,
    random_seed=RANDOM_SEED
)

print(f"Arquitectura de red: {layer_sizes}")
print(f"Función de activación: ReLU")
print(f"Tasa de aprendizaje: 0.01")
print("\nIniciando entrenamiento...")

# Entrenar la red
nn.fit(X_train, y_train, epochs=200, batch_size=16, verbose=True)

print("\nEntrenamiento completado.")
````

````python
# Celda 6: Visualizar la curva de pérdida durante el entrenamiento

plt.figure(figsize=(10, 5))
plt.plot(nn.training_loss, linewidth=2)
plt.xlabel('Epoch')
plt.ylabel('Loss (Cross-Entropy)')
plt.title('Curva de Pérdida Durante el Entrenamiento')
plt.grid()
plt.show()

print(f"Pérdida inicial: {nn.training_loss[0]:.4f}")
print(f"Pérdida final: {nn.training_loss[-1]:.4f}")
````

````python
# Celda 7: Evaluación en conjunto de entrenamiento y test

# Predicciones
y_train_pred = nn.predict(X_train)
y_test_pred = nn.predict(X_test)

# Calcular precisión
train_accuracy = accuracy_score(y_train, y_train_pred)
test_accuracy = accuracy_score(y_test, y_test_pred)

print(f"Precisión en Training: {train_accuracy:.4f}")
print(f"Precisión en Test: {test_accuracy:.4f}")
print("\n" + "="*50)
print("Reporte de Clasificación (Test Set)")
print("="*50)
print(classification_report(y_test, y_test_pred, target_names=['Clase 0', 'Clase 1']))
````

````python
# Celda 8: Matriz de confusión

cm = confusion_matrix(y_test, y_test_pred)

plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
            xticklabels=['Clase 0', 'Clase 1'],
            yticklabels=['Clase 0', 'Clase 1'])
plt.title('Matriz de Confusión - Test Set')
plt.ylabel('Etiqueta Verdadera')
plt.xlabel('Etiqueta Predicha')
plt.show()

print(f"Verdaderos Negativos (TN): {cm[0, 0]}")
print(f"Falsos Positivos (FP): {cm[0, 1]}")
print(f"Falsos Negativos (FN): {cm[1, 0]}")
print(f"Verdaderos Positivos (TP): {cm[1, 1]}")
````

### Interpretación Matemática

La capa oculta de 8 neuronas crea una representación transformada de los datos originales. Cada neurona de la capa oculta calcula:

$$a_j^{[1]} = \text{ReLU}(w_j^T x + b_j)$$

La capa de salida combina linealmente estas características transformadas:

$$\hat{y} = \sigma(w^T a^{[1]} + b)$$

Esta composición de transformaciones no lineales permite que la red capture relaciones complejas en los datos.

## Conclusiones y Recomendaciones

### Conclusiones Clave

1. **Poder Representacional:** Las redes neuronales multicapa pueden aprender fronteras de decisión complejas y no lineales que serían imposibles para modelos lineales.

4. **Necesidad de Regularización:** Con redes más grandes o datos menores, es crucial añadir regularización (L1/L2, dropout) para evitar overfitting.
