## Descripción

Los **grafos** (o redes) son estructuras matemáticas que representan relaciones entre entidades. Están compuestos por **nodos** (vértices) conectados por **aristas** (enlaces), permitiendo modelar sistemas complejos en el mundo real.

La **teoría de grafos** es una rama de las matemáticas discretas que estudia las propiedades, características y comportamientos de estos sistemas de redes. Es fundamental para entender desde redes sociales y sistemas biológicos, hasta infraestructuras de transporte y análisis de datos.

## Estructura del Modelo/Sistema

Un grafo $G = (V, E)$ consta de:

1. **Conjunto de Nodos (V):** entidades o vértices en la red
2. **Conjunto de Aristas (E):** conexiones o relaciones entre nodos
3. **Propiedades de las aristas:**
   - Dirección: aristas dirigidas (→) o no dirigidas (—)
   - Peso: aristas ponderadas con valores (distancia, fortaleza, etc.) o no ponderadas
4. **Atributos de nodos:** características asociadas a cada nodo
5. **Métrica de distancia:** longitud del camino más corto entre nodos

## Modelo Matemático

### Definición Formal de Grafo

Un grafo se define como $G = (V, E)$ donde:
- $V = \{v_1, v_2, \ldots, v_n\}$ es el conjunto de nodos
- $E \subseteq V \times V$ es el conjunto de aristas
- Una arista $e = (u, v)$ conecta los nodos $u$ y $v$

### Métricas Fundamentales

**Grado de un nodo:**
$$\deg(v) = |\{u \in V : (u, v) \in E\}|$$

En un grafo dirigido:
- **In-degree:** aristas entrantes a $v$
- **Out-degree:** aristas salientes de $v$

**Densidad de la red:**
$$\rho = \frac{|E|}{\binom{|V|}{2}} = \frac{2|E|}{|V|(|V|-1)}$$

**Distancia entre nodos:**
$$d(u, v) = \text{longitud del camino más corto de } u \text{ a } v$$

**Diámetro del grafo:**
$$D(G) = \max_{u,v \in V} d(u, v)$$

### Medidas de Centralidad

**Centralidad de Grado:**
$$C_D(v) = \deg(v)$$

**Centralidad de Cercanía (Closeness):**
$$C_C(v) = \frac{n-1}{\sum_{u \neq v} d(v, u)}$$

**Centralidad de Intermediación (Betweenness):**
$$C_B(v) = \sum_{s \neq v \neq t} \frac{\sigma_{st}(v)}{\sigma_{st}}$$
donde $\sigma_{st}$ es el número de caminos más cortos de $s$ a $t$, y $\sigma_{st}(v)$ es el número que pasan por $v$.

**Coeficiente de Agrupamiento (Clustering):**
$$C_{clust}(v) = \frac{|\{e = (u, w) : u, w \in N(v), e \in E\}|}{\binom{|N(v)|}{2}}$$

donde $N(v)$ es el conjunto de vecinos de $v$.

- **Cambios dinámicos:** redes estáticas no capturan evolución temporal

````python
# Celda 1: Importar librerías y configuración
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import networkx as nx
from collections import defaultdict, deque
import seaborn as sns

# Fijar semilla para reproducibilidad
RANDOM_SEED = 42
np.random.seed(RANDOM_SEED)

# Configurar visualizaciones
plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette("husl")

print("Librerías cargadas correctamente.")
print(f"NetworkX version: {nx.__version__}")
print(f"NumPy version: {np.__version__}")
print(f"Pandas version: {pd.__version__}")
````

## Implementación: Conceptos Fundamentales

````python
# Celda 2: Crear un grafo no dirigido simple

# Crear grafo no dirigido
G = nx.Graph()

# Añadir nodos
nodos = ['A', 'B', 'C', 'D', 'E', 'F', 'G']
G.add_nodes_from(nodos)

# Añadir aristas (conexiones)
aristas = [
    ('A', 'B'), ('A', 'C'), ('A', 'D'),
    ('B', 'D'), ('B', 'E'),
    ('C', 'D'), ('C', 'F'),
    ('D', 'E'), ('D', 'F'),
    ('E', 'G'),
    ('F', 'G')
]
G.add_edges_from(aristas)

print(f"Número de nodos: {G.number_of_nodes()}")
print(f"Número de aristas: {G.number_of_edges()}")
print(f"Nodos: {list(G.nodes())}")
print(f"Aristas: {list(G.edges())}")
````

````python
# Celda 3: Visualizar el grafo

plt.figure(figsize=(10, 8))

# Layout usando spring layout para mejor visualización
pos = nx.spring_layout(G, seed=RANDOM_SEED, k=2, iterations=50)

# Dibujar grafo
nx.draw_networkx_nodes(G, pos, node_color='lightblue', node_size=800)
nx.draw_networkx_labels(G, pos, font_size=12, font_weight='bold')
nx.draw_networkx_edges(G, pos, width=2, alpha=0.6)

plt.title('Grafo No Dirigido: Ejemplo Simple', fontsize=14, fontweight='bold')
plt.axis('off')
plt.tight_layout()
plt.show()

print("Grafo visualizado correctamente.")
````

````python
# Celda 4: Análisis de Propiedades Básicas

print("="*60)
print("PROPIEDADES BÁSICAS DEL GRAFO")
print("="*60)

# Grado de cada nodo
print("\n1. GRADO DE CADA NODO:")
print("-" * 40)
grados = dict(G.degree())
for nodo, grado in sorted(grados.items()):
    print(f"  Nodo {nodo}: grado = {grado}")

# Grado promedio
grado_promedio = sum(grados.values()) / len(grados)
print(f"\n  Grado promedio: {grado_promedio:.2f}")

# Densidad
densidad = nx.density(G)
print(f"\n2. DENSIDAD DEL GRAFO: {densidad:.4f}")
print(f"   (Proporción de aristas posibles: {densidad*100:.2f}%)")

# Conectividad
es_conectado = nx.is_connected(G)
print(f"\n3. CONECTIVIDAD: {'Conectado' if es_conectado else 'Desconectado'}")

# Componentes conectadas
num_componentes = nx.number_connected_components(G)
print(f"   Número de componentes: {num_componentes}")

# Diámetro
if es_conectado:
    diametro = nx.diameter(G)
    print(f"\n4. DIÁMETRO: {diametro}")
    print(f"   (Distancia máxima entre dos nodos)")
else:
    print(f"\n4. DIÁMETRO: No aplica (grafo desconectado)")
````

````python
# Celda 5: Medidas de Centralidad

print("\n" + "="*60)
print("MEDIDAS DE CENTRALIDAD")
print("="*60)

# Centralidad de Grado
print("\n1. CENTRALIDAD DE GRADO:")
print("-" * 40)
centralidad_grado = nx.degree_centrality(G)
for nodo in sorted(centralidad_grado.keys(), key=lambda x: centralidad_grado[x], reverse=True):
    print(f"  {nodo}: {centralidad_grado[nodo]:.4f}")

# Centralidad de Cercanía
print("\n2. CENTRALIDAD DE CERCANÍA (Closeness):")
print("-" * 40)
centralidad_cercania = nx.closeness_centrality(G)
for nodo in sorted(centralidad_cercania.keys(), key=lambda x: centralidad_cercania[x], reverse=True):
    print(f"  {nodo}: {centralidad_cercania[nodo]:.4f}")

# Centralidad de Intermediación
print("\n3. CENTRALIDAD DE INTERMEDIACIÓN (Betweenness):")
print("-" * 40)
centralidad_inter = nx.betweenness_centrality(G)
for nodo in sorted(centralidad_inter.keys(), key=lambda x: centralidad_inter[x], reverse=True):
    print(f"  {nodo}: {centralidad_inter[nodo]:.4f}")

# Coeficiente de Agrupamiento
print("\n4. COEFICIENTE DE AGRUPAMIENTO (Clustering):")
print("-" * 40)
clustering = nx.clustering(G)
for nodo in sorted(clustering.keys(), key=lambda x: clustering[x], reverse=True):
    print(f"  {nodo}: {clustering[nodo]:.4f}")

agrupamiento_promedio = nx.average_clustering(G)
print(f"\n  Agrupamiento promedio: {agrupamiento_promedio:.4f}")
````

````python
# Celda 6: Algoritmos de Búsqueda - BFS y DFS

print("\n" + "="*60)
print("ALGORITMOS DE BÚSQUEDA EN GRAFOS")
print("="*60)

# BFS (Breadth-First Search)
print("\n1. BFS (Búsqueda en Amplitud) desde nodo A:")
print("-" * 40)
bfs_order = list(nx.bfs_tree(G, 'A').nodes())
print(f"  Orden de visita: {bfs_order}")

# DFS (Depth-First Search)
print("\n2. DFS (Búsqueda en Profundidad) desde nodo A:")
print("-" * 40)
dfs_order = list(nx.dfs_tree(G, 'A').nodes())
print(f"  Orden de visita: {dfs_order}")

# Árbol BFS
print("\n3. Árbol BFS desde A (aristas de árbol):")
bfs_tree = nx.bfs_tree(G, 'A')
print(f"  Aristas: {list(bfs_tree.edges())}")

# Árbol DFS
print("\n4. Árbol DFS desde A (aristas de árbol):")
dfs_tree = nx.dfs_tree(G, 'A')
print(f"  Aristas: {list(dfs_tree.edges())}")
````

````python
# Celda 7: Caminos más Cortos

print("\n" + "="*60)
print("CAMINOS MÁS CORTOS")
print("="*60)

# Camino más corto entre dos nodos específicos
nodo_inicio = 'A'
nodo_fin = 'G'

camino = nx.shortest_path(G, nodo_inicio, nodo_fin)
longitud = nx.shortest_path_length(G, nodo_inicio, nodo_fin)

print(f"\n1. Camino más corto de {nodo_inicio} a {nodo_fin}:")
print("-" * 40)
print(f"  Camino: {' → '.join(camino)}")
print(f"  Longitud: {longitud}")

# Matriz de distancias entre todos los nodos
print(f"\n2. Matriz de Distancias (caminos más cortos):")
print("-" * 40)

distancias = dict(nx.all_pairs_shortest_path_length(G))
nodos_ordenados = sorted(G.nodes())

# Crear tabla
matriz_dist = pd.DataFrame(
    {nodo: [distancias[nodo][otro] for otro in nodos_ordenados] for nodo in nodos_ordenados},
    index=nodos_ordenados
)
print(matriz_dist)
````

````python
# Celda 8: Crear y analizar un grafo dirigido

print("\n" + "="*60)
print("GRAFO DIRIGIDO")
print("="*60)

# Crear grafo dirigido
G_dir = nx.DiGraph()

# Añadir aristas dirigidas (con dirección)
aristas_dir = [
    ('A', 'B'), ('A', 'C'),
    ('B', 'D'), ('B', 'E'),
    ('C', 'D'), ('C', 'F'),
    ('D', 'E'),
    ('E', 'G'),
    ('F', 'G'),
    ('G', 'A')  # Crear un ciclo
]
G_dir.add_edges_from(aristas_dir)

# Análisis del grafo dirigido
print(f"\nNúmero de nodos: {G_dir.number_of_nodes()}")
print(f"Número de aristas: {G_dir.number_of_edges()}")

print(f"\nIn-degree (aristas entrantes):")
in_degree = dict(G_dir.in_degree())
for nodo in sorted(in_degree.keys()):
    print(f"  {nodo}: {in_degree[nodo]}")

print(f"\nOut-degree (aristas salientes):")
out_degree = dict(G_dir.out_degree())
for nodo in sorted(out_degree.keys()):
    print(f"  {nodo}: {out_degree[nodo]}")

# Detectar ciclos
print(f"\n¿Contiene ciclos?: {not nx.is_directed_acyclic_graph(G_dir)}")

if not nx.is_directed_acyclic_graph(G_dir):
    # Encontrar un ciclo
    ciclo = nx.find_cycle(G_dir)
    print(f"Ciclo encontrado: {ciclo}")
````

````python
# Celda 9: Visualizar grafo dirigido

plt.figure(figsize=(10, 8))

pos_dir = nx.spring_layout(G_dir, seed=RANDOM_SEED, k=2, iterations=50)

# Dibujar grafo dirigido
nx.draw_networkx_nodes(G_dir, pos_dir, node_color='lightcoral', node_size=800)
nx.draw_networkx_labels(G_dir, pos_dir, font_size=12, font_weight='bold')
nx.draw_networkx_edges(G_dir, pos_dir, width=2, alpha=0.6, 
                       edge_color='gray', arrowsize=20, arrowstyle='->')

plt.title('Grafo Dirigido: Ejemplo con Ciclos', fontsize=14, fontweight='bold')
plt.axis('off')
plt.tight_layout()
plt.show()
````

````python
# Celda 10: Grafo Ponderado

print("\n" + "="*60)
print("GRAFO PONDERADO (CON PESOS)")
print("="*60)

# Crear grafo ponderado
G_ponder = nx.Graph()

# Añadir aristas con pesos (simulando distancias)
aristas_ponderadas = [
    ('A', 'B', 4), ('A', 'C', 2),
    ('B', 'D', 5), ('B', 'E', 10),
    ('C', 'D', 8), ('C', 'F', 15),
    ('D', 'E', 2), ('D', 'F', 6),
    ('E', 'G', 3),
    ('F', 'G', 1)
]
G_ponder.add_weighted_edges_from(aristas_ponderadas)

print(f"\nAristas ponderadas (nodo1, nodo2, peso):")
for u, v, w in G_ponder.edges(data=True):
    print(f"  {u} - {v}: peso = {w['weight']}")

# Camino más corto considerando pesos
nodo_a = 'A'
nodo_b = 'G'

camino_ponderado = nx.shortest_path(G_ponder, nodo_a, nodo_b, weight='weight')
longitud_ponderada = nx.shortest_path_length(G_ponder, nodo_a, nodo_b, weight='weight')

print(f"\nCamino más corto de {nodo_a} a {nodo_b} (considerando pesos):")
print(f"  Camino: {' → '.join(camino_ponderado)}")
print(f"  Distancia total: {longitud_ponderada}")

# Comparar con camino no ponderado
camino_no_pond = nx.shortest_path(G_ponder, nodo_a, nodo_b)
print(f"\nCamino más corto (sin considerar pesos):")
print(f"  Camino: {' → '.join(camino_no_pond)}")
print(f"  Longitud (número de aristas): {len(camino_no_pond) - 1}")
````

````python
# Celda 11: Visualizar grafo ponderado

plt.figure(figsize=(12, 8))

pos_pond = nx.spring_layout(G_ponder, seed=RANDOM_SEED, k=2, iterations=50)

# Dibujar nodos
nx.draw_networkx_nodes(G_ponder, pos_pond, node_color='lightgreen', node_size=800)

# Dibujar etiquetas de nodos
nx.draw_networkx_labels(G_ponder, pos_pond, font_size=12, font_weight='bold')

# Dibujar aristas
nx.draw_networkx_edges(G_ponder, pos_pond, width=2, alpha=0.6)

# Dibujar etiquetas de pesos
edge_labels = nx.get_edge_attributes(G_ponder, 'weight')
nx.draw_networkx_edge_labels(G_ponder, pos_pond, edge_labels, font_size=10)

# Destacar el camino más corto
camino_aristas = [(camino_ponderado[i], camino_ponderado[i+1]) for i in range(len(camino_ponderado)-1)]
nx.draw_networkx_edges(G_ponder, pos_pond, edgelist=camino_aristas, 
                       width=4, edge_color='red', alpha=0.8)

plt.title(f'Grafo Ponderado - Camino más corto (A→G) en rojo', 
         fontsize=14, fontweight='bold')
plt.axis('off')
plt.tight_layout()
plt.show()
````

````python
# Celda 12: Detección de Comunidades

print("\n" + "="*60)
print("DETECCIÓN DE COMUNIDADES")
print("="*60)

# Crear un grafo con comunidades claras
G_comm = nx.Graph()

# Comunidad 1
G_comm.add_edges_from([('A1', 'A2'), ('A2', 'A3'), ('A3', 'A1'), 
                        ('A1', 'A4'), ('A4', 'A2')])

# Comunidad 2
G_comm.add_edges_from([('B1', 'B2'), ('B2', 'B3'), ('B3', 'B1'),
                        ('B1', 'B4'), ('B4', 'B3')])

# Comunidad 3
G_comm.add_edges_from([('C1', 'C2'), ('C2', 'C3'), ('C3', 'C1'),
                        ('C1', 'C4'), ('C4', 'C2')])

# Conexiones entre comunidades (débiles)
G_comm.add_edges_from([('A2', 'B2'), ('B3', 'C1')])

print(f"\nGrafo con 3 comunidades creado.")
print(f"Nodos: {G_comm.number_of_nodes()}")
print(f"Aristas: {G_comm.number_of_edges()}")

# Usar algoritmo de Louvain para detección de comunidades
from networkx.algorithms import community

comunidades = list(community.greedy_modularity_communities(G_comm))

print(f"\nComunidades detectadas: {len(comunidades)}")
for i, comm in enumerate(comunidades, 1):
    print(f"  Comunidad {i}: {sorted(list(comm))}")

# Calcular modularidad
modularidad = community.modularity(G_comm, comunidades)
print(f"\nModularidad: {modularidad:.4f}")
print(f"(Rango 0-1: valores altos indican comunidades bien definidas)")
````

````python
# Celda 13: Visualizar Comunidades

plt.figure(figsize=(12, 8))

pos_comm = nx.spring_layout(G_comm, seed=RANDOM_SEED, k=1, iterations=50)

# Crear mapa de colores para comunidades
colores = ['#FF6B6B', '#4ECDC4', '#45B7D1']
color_map = {}

for i, comm in enumerate(comunidades):
    for nodo in comm:
        color_map[nodo] = colores[i % len(colores)]

# Crear lista de colores para los nodos
node_colors = [color_map[nodo] for nodo in G_comm.nodes()]

# Dibujar
nx.draw_networkx_nodes(G_comm, pos_comm, node_color=node_colors, node_size=800)
nx.draw_networkx_labels(G_comm, pos_comm, font_size=10, font_weight='bold')
nx.draw_networkx_edges(G_comm, pos_comm, width=2, alpha=0.5)

plt.title('Detección de Comunidades - Algoritmo Louvain', fontsize=14, fontweight='bold')
plt.axis('off')
plt.tight_layout()
plt.show()

print("Comunidades visualizadas con colores distintos.")
````

## Referencias y Recursos Adicionales

### Referencias Principales

1. **Barabási, A. L. (2016).** *Network Science.* Cambridge University Press.
   - Introducción moderna a redes complejas y teoría de grafos

2. **Diestel, R. (2017).** *Graph Theory (5th ed.).* Springer.
   - Referencia técnica completa de teoría de grafos

3. **Newman, M. E. J. (2010).** *Networks: An Introduction.* Oxford University Press.
   - Enfoque aplicado a redes reales y complejas

4. **Easley, D., & Kleinberg, J. (2010).** *Networks, Crowds, and Markets.* Cambridge University Press.
   - Redes en contextos sociales y económicos

- **Social Network Analysis:** identificar influenciadores, comunidades

- **Internet:** análisis de topología, propagación de contenido
- **Epidemiología:** modelado de contagio de enfermedades
